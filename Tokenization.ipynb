```python
import tiktoken
```

```python
text = "Who is the PM of India ?"
```

```python
tokenizer = tiktoken.encoding_for_model(model_name="gpt-4")
```

```python
tokenids = tokenizer.encode(text)
tokenids
```

```python
resultids = [8971, 315, 28811, 374, 5768, 35973, 13]
result = tokenizer.decode(resultids)
result
```
