```python
## 1. Using OpenAI API

OpenAI provides an API that allows Python applications to communicate with GPT models and generate responses from user prompts.
```

```python
!pip install -U openai
```

```python
from openai import OpenAI
from google.colab import userdata

OPENAI_API_KEY = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
response = client.responses.create(
    model="gpt-5",
    input="Explain Artificial Intelligence in one short paragraph."
)

print(response.output_text)
```

```python
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="what is 5! ?"
)

print(response.text)
```
